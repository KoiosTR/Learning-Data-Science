```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Student Depression Dataset.csv')
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
gender_group = df.groupby('Gender')['Depression'].mean()*100
gender_group
```

```python
gender_count = df['Gender'].value_counts()
gender_count
```

```python
df_analysis = pd.DataFrame('')
```

```python
plt.figure(figsize=(12,6))
plt.bar(gender_group.index,gender_group.values,color=['blue','red'])

plt.title('Depression Rate By Gender',fontsize=20)
plt.xlabel('Gender',fontsize=15)
plt.ylabel('Depression Rate',fontsize=15)
plt.ylim(0,100)
plt.grid(axis='y',linestyle='--',alpha=0.6)
plt.show()
```
